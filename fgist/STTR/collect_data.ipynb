### Перенесем данные в STTR

```python
import os
from tqdm import tqdm
import time

os.system(f"rm -rf /opt/esilvestrov_toolkit_ds/98890/fgist/STTR/inputs/style/*")
os.system(f"rm -rf /opt/esilvestrov_toolkit_ds/98890/fgist/STTR/inputs/content/*")
os.system(f"find /opt/esilvestrov_toolkit_ds/98890/fgist/STTR/inputs/target/ -type f -delete")

for i in tqdm(range(len(train))):
    os.system(f"cp /opt/esilvestrov_toolkit_ds/98890/dataset/teacher_result/style/'{train[i]['style']}' /opt/esilvestrov_toolkit_ds/98890/fgist/STTR/inputs/style/'{train[i]['style']}'")
    os.system(f"cp /opt/esilvestrov_toolkit_ds/98890/dataset/teacher_result/content/'{train[i]['content']}' /opt/esilvestrov_toolkit_ds/98890/fgist/STTR/inputs/content/'{train[i]['content']}'")
    os.system(f"cp /opt/esilvestrov_toolkit_ds/98890/dataset/teacher_result/target/'{train[i]['target']}' /opt/esilvestrov_toolkit_ds/98890/fgist/STTR/inputs/target/'{train[i]['target']}'")
```
