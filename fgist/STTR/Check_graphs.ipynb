```python
import wandb

# Инициализация API
api = wandb.Api()

# Укажите имя пользователя и проекта
username = "decision_forrest_gump"
project_name = "science"

# Получение всех запусков в проекте
runs = api.runs(f"{username}/{project_name}")

# Вывод информации о каждом запуске
for run in runs:
    print(f"Run ID: {run.id}")
    print(f"Name: {run.name}")
    print(f"Full path: {username}/{project_name}/{run.id}")
    print("---")
```

```python
import wandb
import matplotlib.pyplot as plt

# Инициализация API
api = wandb.Api()

# Загрузка данных конкретного запуска (замените на свой путь)
run = api.run("decision_forrest_gump/science/866jgur8")
# Получение истории метрик
history = run.scan_history()

# Собираем данные (например, loss)
steps = []
losses = []

for row in history:
    if "Train_loss" in row:
        steps.append(row["_step"])
        losses.append(row["Train_loss"])

# Построение графика
plt.plot(steps, losses)
plt.xlabel("Step")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.grid()
plt.show()
```
